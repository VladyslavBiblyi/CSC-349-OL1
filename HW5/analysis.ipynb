#1.
####Implicit missing values are values that appear normal but actually mean missing information. Here some medical measurements have zeros that don't make sense, like glucose, blood pressure, and BMI. These zeros can affect how accurately the models make predictions.

####I replaced these zeros with NaN and used SimpleImputer to fill them with the median. I chose the median because it isn't affected much by extreme values. Deleting rows would remove too much data. Imputation allowed me to keep all the records. I also fitted the imputer only on the training data to avoid data leakage.

#2.

####Hard voting chooses the class predicted by most models. For example, if two models predict diabetes and one doesn't, the final prediction is diabetes. Soft voting averages the probabilities from each model instead of just counting votes.

####Here hard voting achieved 73.44% accuracy, while soft voting achieved 72.40%, so hard voting performed slightly better. However, soft voting had better recall for diabetes cases(70% compared to 62%).

####Probability calibration matters because different models can give confidence scores that aren't equally reliable. If one model is too confident, it might influence soft voting too much, even when it's wrong.

#3.

####Out-of-bag evaluation estimates how well a Bagging model performs without needing a separate validation dataset. Bagging trains multiple trees using random samples with replacement, so some records are left out of each tree's training data. These leftover records are used for OOB evaluation.

####Here the OOB accuracy was 75.87% and the test accuracy was 75.00%. Since the results were very close, OOB evaluation gave a good estimate of the model's performance.

####OOB is useful because it saves time compared to traditional cross-validation, where models need to be trained multiple times on different splits.

#4.

####Both Random Forest and Bagging use multiple Decision Trees. The difference is that Bagging uses random samples of the data, while Random Forest also selects random subsets of features at each split.

####Here I used max_features="sqrt", meaning each split only considers a few randomly selected features. This makes the trees less similar, which can reduce variance and help the model perform better on new data.

####My Random Forest accuracy was 75.00%, the same as Bagging. The feature importance graph showed that glucose was the most important feature, followed by BMI and age. Even though Random Forest didn't perform better here, random feature selection can still improve generalization.

####AdaBoost trains models one after another. It starts with equal weights for all training samples, then gives more importance to samples that were predicted incorrectly. This helps the next model focus on the harder cases.

####I used 30 decision stumps with a learning rate of 0.5. My training accuracy was 78.82% and test accuracy was 76.04%. The difference was only 2.78 percentage points, so there weren't strong signs of overfitting.

####AdaBoost also had the highest test accuracy in my project. If it started overfitting, I could reduce the number of estimators, lower the learning rate, or limit the depth of the trees. These changes could help the model perform better on new data.
