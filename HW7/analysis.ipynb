##1.

####Feature scaling is important because the Wine dataset has 13 chemical features with very different ranges. All three clustering algorithms use distances between samples, so features with larger values could have more influence on the results.

####For example, proline ranged from 278 to 1680, and hue ranged from 0.48 to 1.71. Without scaling, proline would dominate many distance calculations, even if other chemical features were important.

####In Task 1, I used StandardScaler to give all features a mean of approximately 0 and a standard deviation of 1. This allowed each feature to contribute more equally to the distance calculations. Scaling was important for K-Means because it uses distances to centroids, and for DBSCAN and HDBSCAN because they identify clusters based on density and distance.

##2.

####I tested K-Means with values of k from 1 to 10. I used the Elbow Method to measure inertia and Silhouette Scores to evaluate how well the clusters were separated.

####The Elbow Method suggested k=3, because the decrease in inertia started slowing down after three clusters. The highest Silhouette Score was also at k=3, indicating that three clusters provided the best separation among the tested values.

####Both methods agreed with the actual number of wine cultivars, which was three. This showed that the chemical features contained enough information to identify the natural groups without using the class labels.

####DBSCAN worked differently because it grouped samples based on density. With eps=2.0 and min_samples=5, I found 5 clusters and 85 noise points. When I changed the parameters to eps=2.5 and min_samples=8, DBSCAN found 2 clusters and 30 noise points.

####HDBSCAN identified 2 clusters and 58 noise points using min_cluster_size=5.

####DBSCAN and HDBSCAN classified some samples as noise because they were located in less dense regions and did not meet the requirements to belong to a cluster.

####K-Means assigned every sample to a group, while DBSCAN and HDBSCAN were more selective. The results also showed that DBSCAN was sensitive to its parameter settings.


##4.

####DBSCAN and HDBSCAN both identify clusters based on density, but they use different approaches.

####DBSCAN relies on two parameters: eps, which controls the distance between neighboring samples, and min_samples, which determines how many nearby samples are needed to form a dense region. Its main limitation is that one eps value is used for the entire dataset.

####I observed this in Task 5. Changing eps from 2.0 to 2.5, along with changing min_samples, reduced the number of clusters from 5 to 2 and noise points from 85 to 30.

####HDBSCAN does not require a fixed eps. Instead, it examines different density levels and selects clusters that remain stable. This makes it more flexible when the data contains groups with different densities.

####In Task 6, HDBSCAN found 2 clusters and 58 noise points. The dendrogram from Task 7 also helped me understand how samples merged into larger groups, which is related to HDBSCAN's hierarchical approach.
