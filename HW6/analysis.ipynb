#####To preserve 95% of the total variance, PCA needed 123 components out of the original 4,096 dimensions. This means we reduced the number of features by about 97% while keeping most of the information.

#####In Task 3, I noticed that using only 10 components caused the faces to lose a lot of detail. The images looked blurry, and smaller features like the eyes, nose, and mouth were harder to distinguish. At 50 and 100 components, the faces became clearer. With 200 components, most details were preserved, and at 300 components, the reconstructed images were very close to the originals.

#####The MSE also decreased as the number of components increased. It went from 0.006622 with 10 components to 0.000107 with 300 components. This shows that adding more components improves reconstruction quality. At around 300 components, the differences became very small, although the reconstruction was not mathematically identical to the original.

#####The first few Eigenfaces capture the most important differences between the face images. Looking at the plots in Task 4, they represent general facial shapes, lighting differences, shadows, and the positions of facial features.

#####The earlier principal components explain more of the total variance because they capture the largest patterns in the dataset. Later components explain smaller amounts of variance and focus on more specific details, such as small changes in facial features, textures, and lighting.

#####This is why the later Eigenfaces start looking more like pixel noise. They represent smaller variations rather than the main structure of the faces. However, these smaller details can still contain useful information.

##3

#####It is important to fit PCA only on the training data to avoid data leakage. Data leakage happens when information from the test set is used during training or preprocessing.

#####PCA does not use target labels, but it still learns patterns and variance from the data. If we fit PCA on the entire dataset before splitting, the test images would influence the principal components used by the model.

#####This could make the test accuracy look better than it actually is because the model would be evaluated using features that were partly learned from the test data.

#####In Task 5, I avoided this problem by splitting the dataset first and fitting PCA only on the training set. I then used the same PCA transformer on the test set. This gives a more reliable evaluation of how the model performs on unseen images.

#####Task 8 showed that PCA can make training much faster without necessarily reducing accuracy. Logistic Regression using raw features achieved 96.25% accuracy and took 31.09 seconds to train. After applying PCA, it achieved the same accuracy but took only 0.94 seconds, including PCA fitting.

#####The RBF SVM was even faster at 0.18 seconds, but its accuracy was lower at 93.75%. The ensemble took 1.45 seconds and achieved 95.00% accuracy. This shows that more complicated models do not always give better results.

#####Even though the accuracy was high in this experiment, PCA and classical classifiers have limitations for complex computer vision tasks. PCA reduces the number of features but does not directly learn spatial relationships between pixels. It also keeps the directions with the most variance, which are not always the most useful for identifying objects.

#####For example, changes in lighting or background can create a lot of variance without helping identify a person. Classical classifiers also have difficulty recognizing objects when their position, size, or angle changes significantly.

#####CNNs are generally better suited for these tasks because they learn local patterns and combine them into more complex features. Based on my results, PCA is useful for compression and improving training efficiency, but it is not a complete solution for complex image recognition problems.
