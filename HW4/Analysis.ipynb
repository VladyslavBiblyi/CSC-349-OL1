##1
####The Heart Disease dataset combines numerical measurements, such as cholesterol and maximum heart rate, with categorical indicators, such as chest pain type and thal. The breast cancer dataset contained only numerical measurements.I scaled numerical features and one-hot encoded categorical features. Scaling matters for the SVC because it uses distances, so larger measurements could otherwise dominate smaller ones. Decision Trees generally do not need scaling because they split features at thresholds based on their order.

##2
####Question marks can hide missing values by being treated as text. The UCI importer already converted them to NaN. My dataset had four missing values in ca and two in thal.I selected median imputation for numerical features because it is less affected by extremes, although these numerical columns had no missing entries. For categorical features, I used the most frequent category. Imputation happens inside the pipeline using training data only. This avoids data leakage, but filling categories with the most common value can overrepresent that category.

##3
####The RBF SVC created a nonlinear boundary using similarities between patients. It can capture complex relationships, but its predictions are harder to explain. A Decision Tree repeatedly splits patients using feature thresholds, creating rules that are easier to follow. My SVC achieved about 89% accuracy and 0.9643 ROC AUC, while the tree achieved about 77% accuracy and 0.8728 ROC AUC. The SVC did better on this test set, but the tree was more interpretable.

##4
####A smaller C brings stronger regularization, which can reduce overfitting but increase underfitting. A larger C penalizes training errors more strongly and can make the model more sensitive to noise. A smaller gamma usually creates a smoother boundary, while a larger gamma creates more local, complex boundaries that may overfit. My five fold selected C = 1 and gamma = 0.01, with a mean validation ROC AUC of 0.8883. The search used training data only.

####My tree’s strongest feature was thal=3, with an importance of 45.8%, than chest pain type=4 with 16.5% and ca=0 with 14.4%. Cholesterol was about 7.3%, ST depression 5.7%, and maximum heart rate 3.8%. These features helped the tree separate patients with and without heart disease. The importance values show how much each feature contributed, not whether it increases or decreases risk. It shows that chest pain, cholesterol and exercise-related measurements are important. For doctors, this tells that studying these indicators together is much better than relying on one. But some require diagnostic testing, so they are not all available during first screening.
