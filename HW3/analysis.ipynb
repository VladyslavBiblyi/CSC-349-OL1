######1. One of the first problems i had was getting the dataset directly from Kaggle. This was my first time trying to load a Kaggle dataset directly into Colab, so I definitely spent some time figuring out how to download it with kagglehub, find the correct file path, and load the CSV correctly.

######Another issue was that TotalCharges looked like a numerical column, but Pandas originally read it as an object. After looking at the data, I found that some entries contained whitespace instead of actual numbers. I used pd.to_numeric() with errors='coerce' to convert the column to numeric and turn those whitespace entries into NaN. This showed 11 missing values in TotalCharges.

######2. I used info() and isnull().sum() to inspect the columns and check for missing values. At first, TotalCharges did not appear to have missing values because the whitespace entries were being treated as text. After converting the column to numeric, I found 11 missing values. The other columns did not show missing values.

######For numerical columns, I used SimpleImputer(strategy='median') because it is less affected by unusually high or low values. For categorical columns, I used SimpleImputer(strategy='most_frequent'). I also removed customerID completely because it is just an identifier and does not provide useful information for predicting churn.

######3. ColumnTransformer allowed me to apply different preprocessing methods to numerical and categorical columns. For numerical features I used median imputation and StandardScaler. For categorical features i used most-frequent imputation and OneHotEncoder. The ColumnTransformer then combined the processed num and cat data.

######The Pipeline connected this directly to the SGDClassifier. This makes the process easier because I can call pipeline.fit() and it automatically performs all preprocessing before training the model. It also helps prevent data leakage because the preprocessing steps are fitted using the training data as part of the pipeline instead of preprocessing the entire dataset before the train/test split.

######4. I used StandardScaler in my final model because that was the scaling method used in the preprocessing pipeline. I did not actually run a separate MinMaxScaler model, so I cannot say that I observed a difference in accuracy, convergence, or ROC-AUC between the two.

######StandardScaler centers numerical features around zero and scales them based on standard deviation, while MinMaxScaler normally scales the values into a fixed range such as 0 to 1. Both can help SGD training because features with very different scales can affect the optimization process. Based on what I actually tested, however, I can only report the results from StandardScaler.

######5. I configured SGDClassifier with loss='log_loss', which maked it work as a logistic regression classifier trained using stochastic gradient descent. Instead of processing the entire training dataset for every model update, SGD updates the model incrementally, which can make training efficient, especially with larger datasets.

######An important advantage of log_loss is that the model supports predict_proba(). I used this to get the probability of each customer belonging to the churn class. This is useful because churn is not always just a simple Yes or No decision. A company could use the probabilities to identify customers with a high risk of leaving.

######ROC-AUC also evaluates the model across different classification thresholds instead of only looking at one cutoff. For example, my 80/20 model had an ROC-AUC of about 0.86, showing that the model had a good ability to distinguish between customers who churned and customers who did not. But my 60/40 split also got AUC = 0.86, so the model's ability to separate churn from non-churn customers remained pretty strong even when you reduced the training set to 60%.

######6. Fiber optic internet and month-to-month contracts had the strongest positive relationship with churn. Fiber optic had the highest positive coefficient at about 1.23, and month-to-month contracts about 1.02. Streaming TV, not having a partner, phone service, and paperless billing also had some positive relationships with churn.

######On the other side, tenure had the strongest negative coefficient at about -1.27, meaning customers who have been with the company longer were less likely to churn. Monthly charges and two-year contracts also had negative coefficients.

######From a business perspective, the company could pay more attention to newer customers, customers with month-to-month contracts, and fiber optic customers because the model associates these features with higher churn.
